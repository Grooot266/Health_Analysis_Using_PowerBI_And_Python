```python
#Defining cramersV funtion 
def cramers_V(var1,var2):
    crosstab =np.array(pd.crosstab(var1,var2, rownames=None , colnames=None))
    stat = chi2_contingency(crosstab)[0]
    obs = np.sum(crosstab)
    mini = min(crosstab.shape)
    return (stat/(obs*mini))
        
```

```python
# Cramer's V statistic for each pair of categorical variables in a DataFrame 
rows=[]
for var1 in df:
    col=[]
    for var2 in df:
        cramers = cramers_V(df[var1],df[var2])
        col.append(round(cramers,2))
    rows.append(col)
cramers_results = np.array(rows)
mat = pd.DataFrame(cramers_results, columns = df.columns, index=df.columns)
```

Implementation of ID3 decision tree algorithm 

```python
from pprint import pprint

# Implement the ID3 algorithm
def entropy(df):
    labels = df.iloc[:, -1]
    unique_labels, label_counts = np.unique(labels, return_counts=True)
    probabilities = label_counts / len(labels)
    return -np.sum(probabilities * np.log2(probabilities))

def information_gain(df, attribute):
    total_entropy = entropy(df)
    values, counts = np.unique(df[attribute], return_counts=True)
    weighted_entropy = np.sum([(counts[i] / np.sum(counts)) * entropy(df[df[attribute] == values[i]]) for i in range(len(values))])
    return total_entropy - weighted_entropy

def id3(df, dt, feature_col, target, parent_node_class=None):
    if len(np.unique(df["physical_health"])) <= 1:
        return np.unique(df["physical_health"])[0]

    # If the dataset is empty, return the most common target value in the original dataset
    elif len(df) == 0:
        return np.unique(df["physical_health"]).argmax()

    # If no features are left, return the parent node class
    elif len(feature_col) == 0:
        return parent_node_class

    else:
        # Choose the best feature to split on
        best_feature = max(feature_col, key=lambda x: information_gain(df, x))
        tree = {best_feature: {}}

        # Remove the best feature from the feature list
        feature_col = [feature for feature in feature_col if feature != best_feature]

        # Grow the tree recursively for each value of the best feature
        for value in np.unique(df[best_feature]):
            sub_data = df.where(df[best_feature] == value).dropna()
            subtree = id3(sub_data, df, feature_col, target, parent_node_class)
            tree[best_feature][value] = subtree

        return tree

# Example usage
# Assuming you have a DataFrame named 'df' with the target attribute named 'target'
features = df.columns[:-1].tolist()
tree = id3(df, df, features, target)
ID3 = id3(df, df, features, target)
pprint(tree)

# Extract features and target attribute name
features = df.columns[:-1].tolist()
target_attribute_name = 'physical_health'

# Build the decision tree
tree = id3(df, df, features, target_attribute_name)

# Display the resulting decision tree
print("\nDecision Tree:")
pprint(tree)

```
